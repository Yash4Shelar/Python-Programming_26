```python
# Program to check whether or not the triangle is right-angled triangle using function.

def check_triangle(x,y,z):
    if x+y<=z or y+z<=x or x+z<=y:
        print(f"{x}, {y}, and {z} are not sides of any triangle.")
    elif pow(x,2) == pow(y,2) + pow(z,2):
        print(f"{x}, {y}, and {z} are sides of a right-angled triangle.")
    elif pow(y,2) == pow(x,2) + pow(z,2):
        print(f"{x}, {y}, and {z} are sides of a right-angled triangle.")
    elif pow(z,2) == pow(x,2) + pow(y,2):
        print(f"{x}, {y}, and {z} are sides of a right-angled triangle.")
    elif x==y!=z or y==z!=x or x==z!=y:
        print(f"{x}, {y}, and {z} are sides of a isosceles triangle.")
    elif x==y==z:
        print(f"{x}, {y}, and {z} are sides of a equilateral triangle.")
    else:
        print(f"{x}, {y}, and {z} are sides of a scalene triangle.")

a = int(input("Enter side a : "))
b = int(input("Enter side b : "))
c = int(input("Enter side c : "))

check_triangle(a,b,c)
```
